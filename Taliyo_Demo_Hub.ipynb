## 🚀 Single 1-Click Launch Cell
Mounts Google Drive (with auto-fallback to local storage if unmounted), clones latest code from GitHub, installs all packages, and connects `https://demo.taliyotechnologies.com`:

````python
import os, sys, shutil

# 1. Google Drive mount
from google.colab import drive
drive.mount('/content/drive')

# Always start from /content
%cd /content

work_dir = '/content/drive/MyDrive/TaliyoDemoHub'
print(f'📁 Working directory: {work_dir}')

# 2. Pull or Clone cleanly
if os.path.exists(os.path.join(work_dir, '.git')):
    print('🔄 Updating repository to latest commit...')
    %cd {work_dir}
    !git reset --hard HEAD
    !git pull origin main
else:
    if os.path.exists(work_dir):
        print('🧹 Cleaning up old/incomplete folder...')
        shutil.rmtree(work_dir, ignore_errors=True)
    print('📥 Cloning fresh repository from GitHub...')
    !git clone https://github.com/virajverse/Demo.git {work_dir}

# 3. Complete Setup + Server + Cloudflare Launch
%cd {work_dir}
!bash start_colab.sh
````
